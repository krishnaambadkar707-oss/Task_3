## 1. Imports & Environment Setup

```python
import sys
import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Scikit-Learn Imports
from sklearn.model_selection import train_test_split, StratifiedKFold, GridSearchCV
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.impute import SimpleImputer
from sklearn.pipeline import Pipeline
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier, GradientBoostingClassifier
from sklearn.metrics import (
    accuracy_score, recall_score, precision_score, f1_score,
    roc_auc_score, average_precision_score, confusion_matrix, roc_curve, precision_recall_curve
)

print(f"Pandas Version: {pd.__version__}")
print(f"NumPy Version: {np.__version__}")
from sklearn import __version__ as sklearn_version
print(f"Scikit-Learn Version: {sklearn_version}")
print("Setup complete. All dependencies loaded successfully.")
```

## 2. Load & Profile Raw Data (FR-01)

```python
# Load raw dataset
raw_data_path = "../data/raw/telco_customer_churn.csv"
df_raw = pd.read_csv(raw_data_path)

print(f"Dataset Dimensions: {df_raw.shape[0]} rows x {df_raw.shape[1]} columns")
print(f"Duplicate Rows: {df_raw.duplicated().sum()}")
print("\n--- Column Data Types ---")
print(df_raw.dtypes)

# Check Churn Distribution
churn_counts = df_raw['Churn'].value_counts()
print("\n--- Churn Class Distribution ---")
for label, count in churn_counts.items():
    print(f"Churn = {label}: {count} ({count/len(df_raw)*100:.1f}%)")

# Check for whitespace blanks in TotalCharges
blank_total_charges = (df_raw['TotalCharges'].astype(str).str.strip() == '').sum()
print(f"\nTotalCharges Blank Spaces (' '): {blank_total_charges} rows detected.")
```

## 3. Data Preprocessing & Data Cleaning (FR-02 & FR-03)

- Convert `TotalCharges` from object/string to float64, converting invalid spaces `' '` into `NaN` and imputing with the median.
- Encode `Churn` target as binary integer: `1` for Yes, `0` for No.
- Remove non-predictive identifier column `customerID` to prevent data leakage.

```python
df_clean = df_raw.copy()

# Fix TotalCharges
df_clean['TotalCharges'] = df_clean['TotalCharges'].astype(str).str.strip()
df_clean['TotalCharges'] = pd.to_numeric(df_clean['TotalCharges'], errors='coerce')
df_clean['TotalCharges'] = df_clean['TotalCharges'].fillna(df_clean['TotalCharges'].median())

# Fix Target
df_clean['Churn'] = df_clean['Churn'].map({'Yes': 1, 'No': 0}).fillna(0).astype(int)

# Drop CustomerID
if 'customerID' in df_clean.columns:
    df_clean = df_clean.drop(columns=['customerID'])

X = df_clean.drop(columns=['Churn'])
y = df_clean['Churn']

print(f"Cleaned Dataset Dimensions: {df_clean.shape[0]} rows x {df_clean.shape[1]} columns")
print(f"TotalCharges dtype after cleaning: {df_clean['TotalCharges'].dtype}")
print(f"Null values in TotalCharges after median imputation: {df_clean['TotalCharges'].isnull().sum()}")
print("Churn target binary distribution:")
print(y.value_counts())
print(f"X shape (Features): {X.shape}")
print(f"y shape (Target): {y.shape}")
```

## 4. Reproducible Stratified Train-Test Split (FR-04)

We use an 80/20 train/test split with `stratify=y` and `random_state=42` to ensure class balance preservation across training and test splits.

```python
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.20, random_state=42, stratify=y
)

print(f"Training set size: {X_train.shape[0]} samples ({y_train.sum()} Churners, {y_train.mean()*100:.1f}% churn rate)")
print(f"Test set size: {X_test.shape[0]} samples ({y_test.sum()} Churners, {y_test.mean()*100:.1f}% churn rate)")
```

## 5. Build Scikit-Learn Preprocessing Pipeline & Model Training (FR-03 & FR-05)

We build an explicit `ColumnTransformer` to handle numerical and categorical features:
- **Numerical**: `StandardScaler()` + `SimpleImputer(strategy='median')`
- **Categorical**: `OneHotEncoder(drop='first', handle_unknown='ignore')` + `SimpleImputer(strategy='most_frequent')`

```python
num_cols = ['tenure', 'MonthlyCharges', 'TotalCharges']
cat_cols = [col for col in X.columns if col not in num_cols]

print(f"Numerical Features: {num_cols}")
print(f"Categorical Features: {cat_cols}")

num_pipeline = Pipeline([
    ('imputer', SimpleImputer(strategy='median')),
    ('scaler', StandardScaler())
])

cat_pipeline = Pipeline([
    ('imputer', SimpleImputer(strategy='most_frequent')),
    ('onehot', OneHotEncoder(drop='first', sparse_output=False, handle_unknown='ignore'))
])

preprocessor = ColumnTransformer([
    ('num', num_pipeline, num_cols),
    ('cat', cat_pipeline, cat_cols)
])

# Instantiate Candidate Model Pipelines
model_logreg = Pipeline([
    ('preprocessor', preprocessor),
    ('classifier', LogisticRegression(class_weight='balanced', random_state=42, max_iter=1000))
])

model_rf = Pipeline([
    ('preprocessor', preprocessor),
    ('classifier', RandomForestClassifier(n_estimators=100, max_depth=8, class_weight='balanced', random_state=42))
])

model_gb = Pipeline([
    ('preprocessor', preprocessor),
    ('classifier', GradientBoostingClassifier(n_estimators=100, learning_rate=0.1, max_depth=4, random_state=42))
])

print("\nFitting Models...")
model_logreg.fit(X_train, y_train)
print("1. Baseline Logistic Regression (Balanced) fitted successfully.")
model_rf.fit(X_train, y_train)
print("2. Random Forest Classifier fitted successfully.")
model_gb.fit(X_train, y_train)
print("3. Gradient Boosting Classifier fitted successfully.")
```

## 6. Comprehensive Model Evaluation & Benchmark Comparison (Section 6 & 12)

```python
def calc_metrics(name, model, X_t, y_t):
    y_pred = model.predict(X_t)
    y_prob = model.predict_proba(X_t)[:, 1]
    acc = accuracy_score(y_t, y_pred)
    rec = recall_score(y_t, y_pred)
    prec = precision_score(y_t, y_pred)
    f1 = f1_score(y_t, y_pred)
    roc = roc_auc_score(y_t, y_prob)
    pr = average_precision_score(y_t, y_prob)
    cm = confusion_matrix(y_t, y_pred)
    tn, fp, fn, tp = cm.ravel()
    return {
        'Model': name,
        'Accuracy': round(acc, 4),
        'Recall': round(rec, 4),
        'Precision': round(prec, 4),
        'F1-Score': round(f1, 4),
        'ROC-AUC': round(roc, 4),
        'PR-AUC': round(pr, 4),
        'TP': tp, 'FP': fp, 'FN': fn, 'TN': tn
    }

results = [
    calc_metrics('Logistic Regression (Bal)', model_logreg, X_test, y_test),
    calc_metrics('Random Forest (Balanced)', model_rf, X_test, y_test),
    calc_metrics('Gradient Boosting / XGB', model_gb, X_test, y_test)
]

df_results = pd.DataFrame(results)
print("="*100)
print("MODEL EVALUATION METRICS COMPARISON TABLE (Test Set, Default Threshold = 0.50)")
print("="*100)
print(df_results.to_string())
print("="*100)
```

## 7. Decision Threshold Trade-off Analysis

In customer churn prediction, **Recall** is frequently more critical than accuracy alone: missing a churner means lost revenue, whereas sending a discount offer to a non-churner incurs only minor campaign cost. We analyze thresholds from `0.10` to `0.90`.

```python
y_prob_gb = model_gb.predict_proba(X_test)[:, 1]
thresholds = np.linspace(0.10, 0.90, 9)
th_list = []

for th in thresholds:
    y_p = (y_prob_gb >= th).astype(int)
    acc = accuracy_score(y_test, y_p)
    rec = recall_score(y_test, y_p)
    prec = precision_score(y_test, y_p, zero_division=0)
    f1 = f1_score(y_test, y_p, zero_division=0)
    tn, fp, fn, tp = confusion_matrix(y_test, y_p).ravel()
    th_list.append({
        'Threshold': round(th, 2),
        'Accuracy': round(acc, 4),
        'Recall': round(rec, 4),
        'Precision': round(prec, 4),
        'F1-Score': round(f1, 4),
        'TP': tp, 'FP': fp, 'FN': fn, 'TN': tn
    })

df_th = pd.DataFrame(th_list)
print("Threshold Impact Analysis on Best Model (Gradient Boosting):")
print(df_th.to_string(index=False))
```

## 8. Feature Importance & Churn Factors Interpretation (Section 7)

Understanding *why* customers churn provides actionable business insights for marketing, product, and retention teams.

```python
# Extract feature names from fitted ColumnTransformer
prep = model_gb.named_steps['preprocessor']
cat_feature_names = prep.named_transformers_['cat'].named_steps['onehot'].get_feature_names_out(cat_cols)
all_feature_names = num_cols + list(cat_feature_names)

importances = model_gb.named_steps['classifier'].feature_importances_
df_imp = pd.DataFrame({
    'Feature': all_feature_names,
    'Importance (%)': np.round(importances * 100, 2)
}).sort_values(by='Importance (%)', ascending=False).reset_index(drop=True)

print("Top 10 Churn Drivers (Gradient Boosting Feature Importance):")
print(df_imp.head(10).to_string(index=False))
```
